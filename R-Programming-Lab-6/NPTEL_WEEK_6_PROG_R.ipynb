## Control structures in R

```r
for (i in 1:5) {print (i)}
```

```r
for (i in 1:5) {print(i^2)}
```

```r
for (i in c(2,4,6,7)) {print(i^2)}
```

```r
x = c(2,4,6,8,10,12)

excount = function(x) {

  count = 0
  for(xval in x){
    if(xval/2 > 3)
    count = count + 1
  }
  print(count)
}
excount(x)
```

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for(x in child){
  for(y in sweet){
    print(paste(x,y))
  }
}
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for(x in drink){
  if(x == "tea"){
    break
  }
  print(x)
}
```

```r
drink = c("coffee", "lemonade", "tea", "juice")

for(x in drink){
  if(x == "lemonade"){
    next
  }
  print(x)
}
```

# THE WHILE LOOP

```r
i = 1
while (i < 10){
  print(i)
  i = i + 1
}
```

```r
i = 1
while(i < 10){
  print(i^2)
  i = i + 2
}
```

```r
sumfunction = function(){
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))

  while(number <= 25){
    sum = sum + number
    number = number + 1
  }

  print(paste("The sum of the numbers received from the while loop: ", sum))
}

sumfunction()
```

# REPEAT

```r
i = 1
repeat{
  print(i)
  i = i + 1
  if(i == 5)
  break
}
```

```r
i = 1
repeat{
  print(i^2)
  i = i + 2
  if(i > 10)
  break
}
```

```r
i = 1
repeat{
  i = i + 1
  if(i<10) next
  print(i^2)
  if(i >=13) break
}
```

# FUNCTIONS

```r
abc = function(x){
  x^2
}

abc(3)
```

```r
abc(6)
```

```r
abc(9)
```

```r
abc = function(x,y){
  x^2+y^2
}

abc(3,4)

abc(10,10)

abc(-2,-3)
```

```r
abc = function(x){
  sin(x)^2+cos(x)^2 + x
}

abc(9)

abc(99)

abc(-15)
```

```r
abc = function(){
  for(i in 1:3){
    print(i^3)
  }
}

abc()
```
